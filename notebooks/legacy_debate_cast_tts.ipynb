
# 🎙️ Debate Cast → MP3 (Offline, pyttsx3)

This notebook generates an MP3 of your **Debate Cast** using **pyttsx3** (offline TTS).  
Works on Ubuntu/Windows/macOS without internet.

### How to use
1. (Optional) Run the **Install** cell if `pyttsx3` is missing.
2. Edit the **Debate Text** cell — paste your debate script.
3. (Optional) Run **List Voices** to see available voices and set one.
4. Run **Generate MP3** — your audio will be saved to `./debate_out/debate_cast.mp3`.


```python

# ⬇️ Install (run if pyttsx3 is missing)
# On Ubuntu, pyttsx3 uses eSpeak by default. If you don't have it:
#   sudo apt-get update && sudo apt-get install -y espeak
try:
    import pyttsx3
    print("pyttsx3 already installed ✓")
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pyttsx3"])
    import pyttsx3
    print("pyttsx3 installed ✓")
```

```python
from pathlib import Path
debate_text = Path("debate_cast_script.txt").read_text(encoding="utf-8")
print(debate_text[:300] + "...")
```

```python

# 🔎 List available voices (optional)
import pyttsx3
engine = pyttsx3.init()
voices = engine.getProperty('voices')
print(f"Found {len(voices)} voices")
for i, v in enumerate(voices):
    print(f"[{i}] id={v.id} | name={getattr(v,'name','')} | lang={getattr(v,'languages', '')} | gender={getattr(v,'gender','?')}")
```

```python

# ⚙️ Settings
VOICE_INDEX = None   # set to an integer from the list above, e.g., 0 or 1. Leave None for default.
RATE_FACTOR = 0.95   # 1.0 is default; <1 slower, >1 faster
VOLUME = 1.0         # 0.0 .. 1.0

from pathlib import Path
OUT_DIR = Path("./debate_out")
OUT_DIR.mkdir(parents=True, exist_ok=True)
OUT_MP3 = OUT_DIR / "debate_cast.mp3"
print("Output will be saved to:", OUT_MP3)
```

```python

# ▶️ Generate MP3
import pyttsx3, os

engine = pyttsx3.init()

# Voice
if VOICE_INDEX is not None:
    voices = engine.getProperty('voices')
    if 0 <= VOICE_INDEX < len(voices):
        engine.setProperty('voice', voices[VOICE_INDEX].id)
        print("Using voice:", voices[VOICE_INDEX].id)
    else:
        print("VOICE_INDEX out of range; using default voice.")

# Rate
rate = engine.getProperty('rate')
engine.setProperty('rate', int(rate * RATE_FACTOR))

# Volume
engine.setProperty('volume', float(VOLUME))

# Save to file (pyttsx3 writes WAV-like; many engines will still accept .mp3 extension.
# If your player complains, change the extension to .wav below.)
engine.save_to_file(debate_text, str(OUT_MP3))
engine.runAndWait()

print("Saved:", OUT_MP3.resolve())
```


### 🛠️ Troubleshooting
- **Ubuntu**: if you get errors, install eSpeak: `sudo apt-get install espeak`  
- If your player can't open the file, change the extension in the notebook from `.mp3` to `.wav` and re-run.
- Use the *List Voices* cell to pick a different voice and set `VOICE_INDEX`.
